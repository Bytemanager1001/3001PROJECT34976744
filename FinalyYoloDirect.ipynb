```python
!pip install -q ultralytics pandas pyyaml
import os
import cv2
import matplotlib.pyplot as plt
DATA_ROOT = "/content/drive/MyDrive/Colab Notebooks/3001"

for folder in ["train", "valid", "test"]:
    path = os.path.join(DATA_ROOT, folder)
    print(folder, os.path.isdir(path), path)

from pathlib import Path
import pandas as pd
import numpy as np
import shutil
import yaml
import torch
from ultralytics import YOLO

BASE_DIR = Path("/content/drive/MyDrive/Colab Notebooks/3001")

TRAIN_DIR = BASE_DIR / "train"
TEST_DIR = BASE_DIR / "test"

# Generated dataset and model outputs
YOLO_DATASET = Path("/content/yolo_dataset")
RUNS_DIR = BASE_DIR / "yolo_original_results"

print("GPU available:", torch.cuda.is_available())
print("Training folder exists:", TRAIN_DIR.exists())
print("Test folder exists:", TEST_DIR.exists())
```

```python
import random

random.seed(42)
np.random.seed(42)

CLASS_NAMES = [
    "loose-meat",
    "twisted-meat",
    "unsealed",
    "wrinkle"
]

# Map CSV class labels to IDs 0–3.
# Adjust aliases below to match the exact labels in your CSV.
CLASS_ALIASES = {
    "loose-meat": 0,
    "loose_meat": 0,
    "loose": 0,
    "twisted-meat": 1,
    "twisted_meat": 1,
    "twisted": 1,
    "unsealed": 2,
    "unsealed packaging": 2,
    "unsealed_packaging": 2,
    "packaging-error": 2,
    "wrinkled": 3,
    "wrinkled_meat": 3,
    "wrinkle": 3
}

IMAGE_EXTS = {".jpg", ".jpeg", ".png", ".bmp", ".webp"}

def find_csv(folder):
    """Finds the single CSV file in a given directory.

    Args:
        folder (Path): The directory to search for a CSV file.

    Returns:
        Path: The path to the found CSV file.

    Raises:
        ValueError: If there is not exactly one CSV file in the directory.
    """
    csvs = list(folder.glob("*.csv"))
    if len(csvs) != 1:
        raise ValueError(
            f"Expected exactly one CSV in {folder}, found: {[p.name for p in csvs]}"
        )
    return csvs[0]

def convert_split(source_dir, split_name, csv_path=None):
    """Converts annotations from a source directory CSV structure into YOLO format.

    Args:
        source_dir (Path): The directory containing the source images and CSV file.
        split_name (str): The name of the split ('train' or 'test').
        csv_path (Path, optional): Path to the CSV file. If None, it will be searched in source_dir.

    Returns:
        None
    """
    csv_path = csv_path or find_csv(source_dir)
    df = pd.read_csv(csv_path)

    required = {
        "filename", "width", "height", "class",
        "xmin", "ymin", "xmax", "ymax"
    }
    missing = required - set(df.columns)
    if missing:
        raise ValueError(f"Missing CSV columns in {csv_path.name}: {missing}")

    # Look up image files by filename.
    image_lookup = {
        p.name: p for p in source_dir.rglob("*")
        if p.is_file() and p.suffix.lower() in IMAGE_EXTS
    }

    # Group all bounding boxes belonging to each image.
    grouped = {
        str(name): group
        for name, group in df.groupby("filename")
    }

    # Keep annotated images together when making a validation split.
    filenames = sorted(grouped.keys())
    random.Random(42).shuffle(filenames)

    if split_name == "train":
        if len(filenames) < 2:
            raise ValueError("Need at least two annotated training images.")
        n_val = max(1, round(0.2 * len(filenames)))
        val_names = set(filenames[:n_val])
        train_names = set(filenames[n_val:])
        assignments = [
            ("train", train_names),
            ("val", val_names)
        ]
    else:
        assignments = [(split_name, set(filenames))]

    for target_split, names in assignments:
        img_out = YOLO_DATASET / "images" / target_split
        lbl_out = YOLO_DATASET / "labels" / target_split
        img_out.mkdir(parents=True, exist_ok=True)
        lbl_out.mkdir(parents=True, exist_ok=True)

        for filename in names:
            if filename not in image_lookup:
                print(f"WARNING: Image not found, skipping: {filename}")
                continue

            group = grouped[filename]
            src = image_lookup[filename]
            dst = img_out / src.name
            shutil.copy2(src, dst)

            # Write one YOLO label file per image.
            lines = []
            for _, row in group.iterrows():
                raw_class = str(row["class"]).strip().lower()
                if raw_class not in CLASS_ALIASES:
                    raise ValueError(
                        f"Unknown class '{row['class']}' in {csv_path.name}. "
                        "Add it to CLASS_ALIASES."
                    )

                class_id = CLASS_ALIASES[raw_class]
                width = float(row["width"])
                height = float(row["height"])

                xmin = float(row["xmin"])
                ymin = float(row["ymin"])
                xmax = float(row["xmax"])
                ymax = float(row["ymax"])

                if width <= 0 or height <= 0:
                    raise ValueError(f"Invalid image dimensions for {filename}")

                # YOLO uses normalised centre coordinates and box dimensions.
                x_center = ((xmin + xmax) / 2) / width
                y_center = ((ymin + ymax) / 2) / height
                box_width = (xmax - xmin) / width
                box_height = (ymax - ymin) / height

                # Check coordinates are valid.
                if not (0 <= x_center <= 1 and 0 <= y_center <= 1
                        and 0 < box_width <= 1 and 0 < box_height <= 1):
                    raise ValueError(
                        f"Invalid bounding box in {filename}: "
                        f"{xmin, ymin, xmax, ymax}"
                    )

                lines.append(
                     f"{class_id} {x_center:.6f} {y_center:.6f} "
                     f"{box_width:.6f} {box_height:.6f}"
                )

            (lbl_out / f"{src.stem}.txt").write_text(
                "\n".join(lines), encoding="utf-8"
            )

    print(f"Converted annotations from {csv_path}")

# Start clean so old generated labels cannot contaminate this run.
if YOLO_DATASET.exists():
    shutil.rmtree(YOLO_DATASET)

convert_split(TRAIN_DIR, "train")
convert_split(TEST_DIR, "test")

# Write YOLO dataset configuration.
data_config = {
    "path": str(YOLO_DATASET),
    "train": "images/train",
    "val": "images/val",
    "test": "images/test",
    "names": {i: name for i, name in enumerate(CLASS_NAMES)}
}

config_path = YOLO_DATASET / "data.yaml"
with open(config_path, "w") as f:
    yaml.safe_dump(data_config, f, sort_keys=False)

print("\nDataset configuration:")
print(config_path.read_text())
```

```python
model = YOLO("yolov8n.pt")

train_results = model.train(
    data=str(config_path),
    epochs=50,
    imgsz=640,
    batch=16,
    patience=10,
    project=str(RUNS_DIR),
    name="original_images",
    exist_ok=True,
    plots=True,
    seed=42
)
```

```python
best_model_path = RUNS_DIR / "original_images" / "weights" / "best.pt"
best_model = YOLO(str(best_model_path))

test_results = best_model.val(
    data=str(config_path),
    split="test",
    imgsz=640,
    batch=16,
    plots=True,
    project=str(RUNS_DIR),
    name="original_test",
    exist_ok=True
)

print("\nTest results")
print(f"mAP50:     {test_results.box.map50 * 100:.2f}%")
print(f"mAP50-95:  {test_results.box.map * 100:.2f}%")
print(f"Precision: {test_results.box.mp * 100:.2f}%")
print(f"Recall:    {test_results.box.mr * 100:.2f}%")
```

```python
from pathlib import Path
import pandas as pd

# Fallback definition in case setup cells haven't been run in this session
if 'TEST_DIR' not in globals():
    BASE_DIR = Path("/content/drive/MyDrive/Colab Notebooks/3001")
    TEST_DIR = BASE_DIR / "test"

class_names = [
    "Loose meat",
    "Twisted meat",
    "Unsealed packaging",
    "Wrinkled meat"
]

CONF_THRESHOLD = 0.25

# ------------------------------------------------------------
# 1. PER-DEFECT RECALL
# ------------------------------------------------------------

print("\nPercentage correctly detected for each defect category")
print("-" * 55)

# Note: Make sure to execute the cell containing 'test_results = best_model.val(...)' first!
if 'test_results' not in globals():
    print("WARNING: 'test_results' is not defined in the current session. Please run the cell above to compute validation results.")
else:
    for i, class_name in enumerate(class_names):
        if (
            hasattr(test_results, "ap_class_index")
            and i in list(test_results.ap_class_index)
        ):
            eval_idx = list(test_results.ap_class_index).index(i)
            recall = test_results.box.r[eval_idx] * 100
            print(f"{class_name}: {recall:.2f}%")
        else:
            print(f"{class_name}: N/A (No evaluated test instances)")


# ------------------------------------------------------------
# 2. PACKAGE-LEVEL FAULT DETECTION ACCURACY
# ------------------------------------------------------------
# Ground truth:
#   Faulty     = CSV contains at least one defect annotation
#   Error-free = CSV has no defect annotation for that image
#
# Prediction:
#   Faulty     = YOLO predicts at least one defect above threshold
#   Error-free = YOLO predicts no defects above threshold
#
# Note: TEST_DIR contains the original test images and
# exactly one CSV annotation file.

def find_csv(folder):
    csv_files = list(folder.glob("*.csv"))

    if len(csv_files) != 1:
        raise ValueError(
            f"Expected one CSV in {folder}, found {len(csv_files)}"
        )

    return csv_files[0]

csv_path = find_csv(TEST_DIR)
df = pd.read_csv(csv_path)

required_columns = {"filename", "class"}
if not required_columns.issubset(df.columns):
    raise ValueError(f"CSV must contain {required_columns}")

# An image is faulty if it has one or more annotated defects.
df["image_name"] = df["filename"].astype(str).map(
    lambda x: Path(x.strip()).name.lower()
)

ground_truth = (
    df.groupby("image_name")["class"]
    .apply(lambda classes: classes.notna().any())
    .to_dict()
)

# Include error-free images in the test set.
# Images with no CSV annotation row are treated as error-free,
# so the CSV must represent all test images for this to be valid.
test_images = [
    p for p in TEST_DIR.iterdir()
    if p.is_file() and p.suffix.lower() in {".jpg", ".jpeg", ".png"}
]

correct = 0
tp_faulty = 0
tn_error_free = 0
fp_error_free = 0
fn_faulty = 0
evaluated = 0

for image_path in test_images:
    image_name = image_path.name.lower()

    # Ground-truth package status
    actual_faulty = ground_truth.get(image_name, False)

    # Predict defects in the image
    if 'best_model' not in globals():
        print("ERROR: 'best_model' is not defined. Please run the model loading cell above first.")
        break

    results = best_model.predict(
        source=str(image_path),
        conf=CONF_THRESHOLD,
        verbose=False
    )

    predicted_faulty = (
        results[0].boxes is not None
        and len(results[0].boxes) > 0
    )

    evaluated += 1

    if actual_faulty == predicted_faulty:
        correct += 1

    if actual_faulty and predicted_faulty:
        tp_faulty += 1
    elif not actual_faulty and not predicted_faulty:
        tn_error_free += 1
    elif not actual_faulty and predicted_faulty:
        fp_error_free += 1
    else:
        fn_faulty += 1

if evaluated > 0:
    package_accuracy = 100 * correct / evaluated

    print("\nPACKAGE-LEVEL FAULT DETECTION")
    print("-" * 55)
    print(f"Confidence threshold: {CONF_THRESHOLD}")
    print(f"Images evaluated: {evaluated}")
    print(f"Correct package decisions: {correct}")
    print(f"Package-level accuracy: {package_accuracy:.2f}%")

    print("\nBreakdown")
    print(f"Faulty packages correctly identified: {tp_faulty}")
    print(f"Error-free packages correctly identified: {tn_error_free}")
    print(f"False alarms on error-free packages: {fp_error_free}")
    print(f"Faulty packages missed: {fn_faulty}")
else:
    print("\nNo test images were evaluated. Please check that TEST_DIR contains valid images.")
```
